Additional info: Over a century ago, an Italian economist named Vilfredo Pareto discovered that roughly 20% of the customers account for 80% of the typical retail sales. This is called the [Pareto principle](https://en.wikipedia.org/wiki/Pareto_principle). Check if this dataset displays this characteristic.

```python
customers.info()
```

```python
customers.describe()
```

```python
sns.pairplot(customers)
```

```python
numeric_cols = customers.select_dtypes(include='number').columns

pareto_results = {}

for col in numeric_cols:
    agg = (
        customers.groupby(['Channel', 'Region'])[col]
          .sum()
          .sort_values(ascending=False)
    )
    pareto_results[col] = agg.cumsum().iloc[0] / agg.sum()

print(pd.Series(pareto_results, name='top_unit_share'))

for category in ['Channel', 'Region']:
    for element in numeric_cols:
        agg = (
            customers.groupby('Channel')[element]
            .sum()
            .sort_values(ascending=False)
                )
    
        acummulated_share = agg.cumsum() / agg.sum()
    
        pareto = pd.DataFrame({
                                str(element): agg,
                                'acummulated_share': acummulated_share
                                })
    
        top_fraction = int(0.2 * len(pareto)) or 1
        explained = pareto.iloc[top_fraction - 1]['acummulated_share']
    
        print(f"{element} explains {explained:.1%} of the {category}")
```

**Your observations here**

- ex.: Frozen, Grocery, Milk and Detergents Paper have a high...
- Fresh, Frozen, and Detergent_Paper shows basically a typical Pareto for both Channel and Region when looked individually, but when we look at them as a whole, they are barely more than 50%

# Challenge 3 - Data Preprocessing

One problem with the dataset is the value ranges are remarkably different across various categories (e.g. `Fresh` and `Grocery` compared to `Detergents_Paper` and `Delicassen`). If you made this observation in the first challenge, you've done a great job! This means you not only completed the bonus questions in the previous Supervised Learning lab but also researched deep into [*feature scaling*](https://en.wikipedia.org/wiki/Feature_scaling). Keep on the good work!

Diverse value ranges in different features could cause issues in our clustering. The way to reduce the problem is through feature scaling. We'll use this technique again with this dataset.

#### We will use the `StandardScaler` from `sklearn.preprocessing` and scale our data. Read more about `StandardScaler` [here](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler).

*After scaling your data, assign the transformed data to a new variable `customers_scale`.*

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_cols = customers.select_dtypes(include='number').columns

customers_scale = scaler.fit_transform(customers[numeric_cols])

customers_scale
```

# Challenge 4 - Data Clustering with K-Means

Now let's cluster the data with K-Means first. Initiate the K-Means model, then fit your scaled data. In the data returned from the `.fit` method, there is an attribute called `labels_` which is the cluster number assigned to each data record. What you can do is to assign these labels back to `customers` in a new column called `customers['labels']`. Then you'll see the cluster results of the original data.

```python
from sklearn.cluster import KMeans

k_means = KMeans(init = "k-means++", n_clusters = 4, n_init = 12).fit(customers_scale)
k_means_cluster_centers = k_means.cluster_centers_
customers['labels'] = k_means.labels_
customers
```

### Looking to the elbow we can choose 2 like the correct number of clusters

```python
kmeans_2 = KMeans(n_clusters=2).fit(customers_scale)

labels = kmeans_2.predict(customers_scale)

clusters = kmeans_2.labels_.tolist()
```

```python
customers['Label'] = clusters
```

Count the values in `labels`.

```python
print(customers['Label'].nunique())
print(customers['labels'].nunique())
```

# Challenge 5 - Data Clustering with DBSCAN

Now let's cluster the data using DBSCAN. Use `DBSCAN(eps=0.5)` to initiate the model, then fit your scaled data. In the data returned from the `.fit` method, assign the `labels_` back to `customers['labels_DBSCAN']`. Now your original data have two labels, one from K-Means and the other from DBSCAN.

```python
from sklearn.cluster import DBSCAN 

classifier = DBSCAN(eps=0.5).fit(customers_scale)
customers['labels_DBSCAN'] = classifier.labels_
```

Count the values in `labels_DBSCAN`.

```python
print(customers['labels_DBSCAN'].nunique())
```

# Challenge 6 - Compare K-Means with DBSCAN

Now we want to visually compare how K-Means and DBSCAN have clustered our data. We will create scatter plots for several columns. For each of the following column pairs, plot a scatter plot using `labels` and another using `labels_DBSCAN`. Put them side by side to compare. Which clustering algorithm makes better sense?

Columns to visualize:

* `Detergents_Paper` as X and `Milk` as y
* `Grocery` as X and `Fresh` as y
* `Frozen` as X and `Delicassen` as y

Visualize `Detergents_Paper` as X and `Milk` as y by `labels` and `labels_DBSCAN` respectively

```python
def plot(x,y,hue):
    sns.scatterplot(x=x, 
                    y=y,
                    hue=hue)
    plt.title('Detergents Paper vs Milk ')
    return plt.show();
```

```python
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['Label'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_DBSCAN'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels'])
```

Visualize `Grocery` as X and `Fresh` as y by `labels` and `labels_DBSCAN` respectively

```python
plot(customers['Grocery'],customers['Fresh'], hue=customers['Label'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_DBSCAN'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels'])
```

Visualize `Frozen` as X and `Delicassen` as y by `labels` and `labels_DBSCAN` respectively

```python
plot(customers['Frozen'],customers['Delicassen'], hue=customers['Label'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_DBSCAN'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels'])
```

Let's use a groupby to see how the mean differs between the groups. Group `customers` by `labels` and `labels_DBSCAN` respectively and compute the means for all columns.

```python
customers[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen','Label', 'labels','labels_DBSCAN']].groupby(by=['Label', 'labels','labels_DBSCAN']).mean()
```

```python
means_Label = customers[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen','Label']].groupby('Label').mean()
means_Label
```

```python
means_labels  = customers[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen','labels']].groupby('labels').mean()
means_labels
```

```python
means_labels_DBSCAN = customers[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen','labels_DBSCAN']].groupby('labels_DBSCAN').mean()
means_labels_DBSCAN
```

```python
sep_Label = means_Label.var()
sep_labels = means_labels.var()
sep_dbscan = means_labels_DBSCAN.var()
```

```python
comparison = (
    pd.DataFrame({
        'labels_sep': sep_labels,
        'Label_sep': sep_Label,
        'dbscan_sep': sep_dbscan
    })
    .sort_values(by='dbscan_sep', ascending=False)
)

comparison
```

Which algorithm appears to perform better?

**Your observations here**

- The K-means with 4 clusters (labels) separates Fresh and Milk better than DBSCAN (labels_DBSCAN); however, the other 4 variables DBSCAN separates better.
- The K-means with 4 clusters (labels) separates Fresh and Delicassen better than the K-means with 2 clusters (Label), the other 4 variables are better separated by the 2-cluster one.
- The DBSCAN (labels_DBSCAN) separates Detergent_Paper and Delicassen better than the K-means with 2 clusters (Label), the other 4 variables are better separated by the K-means with 2 clusters (Label).
- Overall, K-means with 2 clusters is better due to its separation magnitude.

# Bonus Challenge 2 - Changing K-Means Number of Clusters

As we mentioned earlier, we don't need to worry about the number of clusters with DBSCAN because it automatically decides that based on the parameters we send to it. But with K-Means, we have to supply the `n_clusters` param (if you don't supply `n_clusters`, the algorithm will use `8` by default). You need to know that the optimal number of clusters differs case by case based on the dataset. K-Means can perform badly if the wrong number of clusters is used.

In advanced machine learning, data scientists try different numbers of clusters and evaluate the results with statistical measures (read [here](https://en.wikipedia.org/wiki/Cluster_analysis#External_evaluation)). We are not using statistical measures today but we'll use our eyes instead. In the cells below, experiment with different number of clusters and visualize with scatter plots. What number of clusters seems to work best for K-Means?

```python
k_means_6 = KMeans(init = "k-means++", n_clusters = 6, n_init = 12).fit(customers_scale)
k_means_5 = KMeans(init = "k-means++", n_clusters = 5, n_init = 12).fit(customers_scale)
k_means_4 = KMeans(init = "k-means++", n_clusters = 4, n_init = 12).fit(customers_scale)
k_means_3 = KMeans(init = "k-means++", n_clusters = 3, n_init = 12).fit(customers_scale)
k_means_2 = KMeans(init = "k-means++", n_clusters = 2, n_init = 12).fit(customers_scale)

customers['labels_6'] = k_means_6.labels_.tolist()
customers['labels_5'] = k_means_5.labels_.tolist()
customers['labels_4'] = k_means_4.labels_.tolist()
customers['labels_3'] = k_means_3.labels_.tolist()
customers['labels_2'] = k_means_2.labels_.tolist()
customers
```

```python
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_6'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_5'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_4'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_3'])
plot(customers['Detergents_Paper'],customers['Milk'], hue=customers['labels_2'])
```

```python
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_6'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_5'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_4'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_3'])
plot(customers['Grocery'],customers['Fresh'], hue=customers['labels_2'])
```

```python
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_6'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_5'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_4'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_3'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_2'])
```

# Bonus Challenge 3 - Changing DBSCAN `eps` and `min_samples`

Experiment changing the `eps` and `min_samples` params for DBSCAN. See how the results differ with scatter plot visualization.

```python
from sklearn.cluster import DBSCAN 

classifier = DBSCAN(eps=0.8, min_samples=15).fit(customers_scale)
customers['labels_DBSCAN_var_eps'] = classifier.labels_
print(customers['labels_DBSCAN_var_eps'].nunique())

plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_DBSCAN_var_eps'])
plot(customers['Frozen'],customers['Delicassen'], hue=customers['labels_DBSCAN'])
```

**Your comment here**

- I can see the number of clusters changed with this, this mixed with the elbow method can be extremely powerful.
