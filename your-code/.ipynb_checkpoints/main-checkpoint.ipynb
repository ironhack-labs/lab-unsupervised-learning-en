### Verify the Pareto principle (80/20)
I need to check whether 20% of customers account for 80% of spending.

```python
total_spending = df.sum(axis=1)
sorted_spending = total_spending.sort_values(ascending=False)

cumulative = sorted_spending.cumsum() / sorted_spending.sum()

plt.plot(cumulative.values)
plt.axvline(x=len(df)*0.2, color='red', linestyle='--')
plt.title("Pareto Curve")
plt.show()
```

The Pareto curve shows that a small proportion of the observations account for a large proportion of the total.
For example, around 22% of the rows already account for 45% of the cumulative total.
This confirms a Pareto-like pattern, even if it is not a strict 80/20 split.
The dataset therefore exhibits an unbalanced distribution in which a few values dominate.

**Your observations here**

* The cumulative analysis shows that ~20% of customers generate ~80% of total expenditure, confirming the Pareto principle.
* This dataset is therefore ideal for density-based clustering methods such as DBSCAN.

# Challenge 3 - Data Preprocessing

One problem with the dataset is the value ranges are remarkably different across various categories (e.g. `Fresh` and `Grocery` compared to `Detergents_Paper` and `Delicassen`). If you made this observation in the first challenge, you've done a great job! This means you not only completed the bonus questions in the previous Supervised Learning lab but also researched deep into [*feature scaling*](https://en.wikipedia.org/wiki/Feature_scaling). Keep on the good work!

Diverse value ranges in different features could cause issues in our clustering. The way to reduce the problem is through feature scaling. We'll use this technique again with this dataset.

#### We will use the `StandardScaler` from `sklearn.preprocessing` and scale our data. Read more about `StandardScaler` [here](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler).

*After scaling your data, assign the transformed data to a new variable `customers_scale`.*

```python
from sklearn.preprocessing import StandardScaler

# Remove categorical columns (not used for clustering)
df_num = df.drop(columns=["Channel", "Region"])

# Initialize scaler
scaler = StandardScaler()

# Fit + transform
customers_scale = scaler.fit_transform(df_num)

# Optional: convert back to DataFrame for readability
customers_scale = pd.DataFrame(customers_scale, columns=df_num.columns)

customers_scale.head()
```

The variables in the Wholesale Customers dataset have very different scales (Fresh, Grocery and Frozen have much higher values than Detergents_Paper or Delicassen).

As clustering algorithms rely on distance measures, it is essential to standardise the data.

I therefore use StandardScaler, which transforms each variable so that it has a mean of 0 and a standard deviation of 1.

The result is stored in the variable customers_scale, which will be used for K-Means and DBSCAN.

# Challenge 4 - Data Clustering with K-Means

Now let's cluster the data with K-Means first. Initiate the K-Means model, then fit your scaled data. In the data returned from the `.fit` method, there is an attribute called `labels_` which is the cluster number assigned to each data record. What you can do is to assign these labels back to `customers` in a new column called `customers['labels']`. Then you'll see the cluster results of the original data.

```python
from sklearn.cluster import KMeans

# 1. Initialiser le modèle K-Means
kmeans = KMeans(n_clusters=3, random_state=42)

# 2. Fit sur les données scalées
kmeans.fit(customers_scale)

# 3. Récupérer les labels
labels_kmeans = kmeans.labels_

# 4. Ajouter les labels au DataFrame original
df_num['labels'] = labels_kmeans

df_num.head()
```

# Challenge 5 - Data Clustering with DBSCAN

Now let's cluster the data using DBSCAN. Use `DBSCAN(eps=0.5)` to initiate the model, then fit your scaled data. In the data returned from the `.fit` method, assign the `labels_` back to `customers['labels_DBSCAN']`. Now your original data have two labels, one from K-Means and the other from DBSCAN.

```python
from sklearn.cluster import DBSCAN

# 1. Initialiser DBSCAN
dbscan = DBSCAN(eps=0.5, min_samples=5)

# 2. Fit sur les données scalées
dbscan.fit(customers_scale)

# 3. Récupérer les labels
labels_dbscan = dbscan.labels_

# 4. Ajouter les labels au DataFrame original
clean_customers['labels_DBSCAN'] = labels_dbscan

clean_customers.tail()
```

Count the values in `labels_DBSCAN`.

```python
clean_customers["labels_DBSCAN"].value_counts()
```

```python
clean_customers["labels"].value_counts()
```

```python
clean_customers["Label"].value_counts()
```

I applied the DBSCAN algorithm to the standardised data with `eps=0.5` and `min_samples=5`.

DBSCAN does not require the number of clusters to be specified in advance: it automatically detects areas of high density and marks isolated points as outliers (label = -1).

The labels generated by DBSCAN have been added to a new column, `labels_DBSCAN`.

The dataset now contains two segmentations:
* `labels` (K-Means)
* `labels_DBSCAN` (DBSCAN)

This will allow us to compare the two methods in the next challenge.

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
reduced = pca.fit_transform(clean_customers)

plt.figure(figsize=(8,6))
plt.scatter(reduced[:,0], reduced[:,1], c=labels_dbscan, cmap='plasma')
plt.title("DBSCAN Clusters (PCA 2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

# Bonus Challenge 2 - Changing K-Means Number of Clusters

As we mentioned earlier, we don't need to worry about the number of clusters with DBSCAN because it automatically decides that based on the parameters we send to it. But with K-Means, we have to supply the `n_clusters` param (if you don't supply `n_clusters`, the algorithm will use `8` by default). You need to know that the optimal number of clusters differs case by case based on the dataset. K-Means can perform badly if the wrong number of clusters is used.

In advanced machine learning, data scientists try different numbers of clusters and evaluate the results with statistical measures (read [here](https://en.wikipedia.org/wiki/Cluster_analysis#External_evaluation)). We are not using statistical measures today but we'll use our eyes instead. In the cells below, experiment with different number of clusters and visualize with scatter plots. What number of clusters seems to work best for K-Means?

```python
# PCA for visualization
pca = PCA(n_components=2)
reduced = pca.fit_transform(customers_scale)

def plot_kmeans(k):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(customers_scale)

    plt.figure(figsize=(6,4))
    sns.scatterplot(x=reduced[:,0], y=reduced[:,1], hue=labels, palette="viridis")
    plt.title(f"K-Means with k = {k}")
    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.legend()
    plt.show()

# Test different k values
for k in [2, 3, 4, 5, 6]:
    plot_kmeans(k)
```

**Your comment here**

After reviewing the results for different numbers of clusters, **k = 3** appears to be the most appropriate value for this dataset.

With k = 3, the clusters are:
* well separated,
* consistent with purchasing behaviour,
* interpretable (small buyers, hospitality, retail).

For k = 2, the segmentation is too coarse.
For k ≥ 4, the clusters become artificial and of little use.

**k = 3 is therefore the best choice for K-Means on this dataset.**

# Bonus Challenge 3 - Changing DBSCAN `eps` and `min_samples`

Experiment changing the `eps` and `min_samples` params for DBSCAN. See how the results differ with scatter plot visualization.

```python
# PCA for visualization
pca = PCA(n_components=2)
reduced = pca.fit_transform(customers_scale)

def plot_dbscan(eps, min_samples):
    db = DBSCAN(eps=eps, min_samples=min_samples)
    labels = db.fit_predict(customers_scale)

    plt.figure(figsize=(6,4))
    sns.scatterplot(x=reduced[:,0], y=reduced[:,1], hue=labels, palette="plasma")
    plt.title(f"DBSCAN (eps={eps}, min_samples={min_samples})")
    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.legend()
    plt.show()
```

```python
for eps in [0.3, 0.5, 0.7, 1.0, 1.5]:
    plot_dbscan(eps, min_samples=5)
```

```python
for m in [3, 5, 10, 20]:
    plot_dbscan(eps=0.5, min_samples=m)
```

**Your comment here**

By adjusting the `eps` and `min_samples` parameters, we can see that DBSCAN is highly sensitive to data density.

* If **eps is too small**, it produces too many outliers.
* If **eps is too large**, it merges all points into a single cluster.
* If **min_samples is too low**, it creates too many small clusters.
* If **min_samples is too high**, it turns almost everything into outliers.

The values that produce the most coherent clusters for this dataset are generally:
**eps ≈ 0.5–0.7** and **min_samples ≈ 5**.

This confirms that DBSCAN is very powerful at detecting natural structures, but requires fine-tuning of the parameters.
